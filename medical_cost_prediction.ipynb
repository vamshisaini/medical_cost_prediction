```python
import pandas as pd
import numpy as np
import sklearn
```

```python
df = pd.read_csv('medical_cost_prediction_dataset.csv')
df.head()
```

```python
df.duplicated().sum()
```

```python
df['insurance_type'] = df['insurance_type'].fillna('unknown')
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
for i in df.columns:
    print('*'*20,i,'*'*20)
    print(df[i].unique())
    print()
```

```python
X=df.drop('annual_medical_cost',axis=1)
y=df[['annual_medical_cost']]
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,train_size =0.8,random_state=42)
```

```python
X_train
```

```python
for i in [X_train,X_test,y_train,y_test]:
    i.reset_index(drop=True,inplace=True)
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
cat_train=X_train.select_dtypes(include='object')

num_train = X_train[['age','bmi','daily_steps','sleep_hours','stress_level','doctor_visits_per_year','insurance_coverage_pct','previous_year_cost']]

no_train = X_train[['diabetes','hypertension','heart_disease','asthma','hospital_admissions','medication_count']]
```

```python
from sklearn.preprocessing import OneHotEncoder

OHE = OneHotEncoder(sparse_output = False).set_output(transform ='pandas')
cat_train_trans = OHE.fit_transform(cat_train)
```

```python
cat_train_trans
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler().set_output(transform ='pandas')

num_train_trans = scaler.fit_transform(num_train)
```

```python
num_train_trans
```

```python
X_train_trans = pd.concat([cat_train_trans,num_train_trans,no_train],axis = 1)
```

```python
X_train_trans
```

# Train model using algorithm

```python
from sklearn.neighbors import KNeighborsRegressor

knn = KNeighborsRegressor()

knn.fit(X_train_trans,y_train)
```

## step --> 5 Feature Engineering on X_test

```python
cat_test = X_test.select_dtypes(include='object')
cat_test
```

```python
cat_test_trans = OHE.transform(cat_test)
cat_test_trans
```

```python
num_test = X_test[['age','bmi','daily_steps','sleep_hours','stress_level','doctor_visits_per_year','insurance_coverage_pct','previous_year_cost']]
num_test_trans = scaler.transform(num_test)
num_test_trans
```

```python
no_trans_test = X_test[['diabetes','hypertension','heart_disease','asthma','hospital_admissions','medication_count']]
```

```python
X_test_trans = pd.concat([cat_test_trans,num_test_trans,no_trans_test],axis=1)
X_test_trans
```

## Step 6 --> prediction on X_test

```python
y_pred = knn.predict(X_test_trans)
```

## step 7 --> Passing y_test and y_pred to evaluation metrics

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error,root_mean_squared_error
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error,root_mean_squared_error
print('r2_score:',r2_score(y_pred,y_test))
print('MAE:',mean_absolute_error(y_pred,y_test))
print('MSE:',mean_squared_error(y_pred,y_test))
print('RMSE:',root_mean_squared_error(y_pred,y_test))
```

### decision tree

```python
from sklearn.tree import DecisionTreeRegressor
dt = DecisionTreeRegressor()

dt.fit(X_train_trans,y_train)
```

```python
y_pred_dt = dt.predict(X_test_trans)

print('r2_score:',r2_score(y_pred_dt,y_test))
print('MAE:',mean_absolute_error(y_pred_dt,y_test))
print('MSE:',mean_squared_error(y_pred_dt,y_test))
print('RMSE:',root_mean_squared_error(y_pred_dt,y_test))
```

### Linear Regression

```python
from sklearn.linear_model import LinearRegression 

LR = LinearRegression()

LR.fit(X_train_trans,y_train)
```

```python
y_pred_lr = LR.predict(X_test_trans)

print('r2_score:',r2_score(y_pred_lr,y_test))
print('MAE:',mean_absolute_error(y_pred_lr,y_test))
print('MSE:',mean_squared_error(y_pred_lr,y_test))
print('RMSE:',root_mean_squared_error(y_pred_lr,y_test))
```

### Random Forest

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor()

rf.fit(X_train_trans,y_train)
```

```python
y_pred_rf = rf.predict(X_test_trans)

print('r2_score:',r2_score(y_pred_rf,y_test))
print('MAE:',mean_absolute_error(y_pred_rf,y_test))
print('MSE:',mean_squared_error(y_pred_rf,y_test))
print('RMSE:',root_mean_squared_error(y_pred_rf,y_test))
```

### xgboost

```python
from xgboost import XGBRegressor

xgb = XGBRegressor()

xgb.fit(X_train_trans,y_train)
```

```python
y_pred_xgb = xgb.predict(X_test_trans)

print('r2_score:',r2_score(y_pred_xgb,y_test))
print('MAE:',mean_absolute_error(y_pred_xgb,y_test))
print('MSE:',mean_squared_error(y_pred_xgb,y_test))
print('RMSE:',root_mean_squared_error(y_pred_xgb,y_test))
```
